# Train and Test split
Now with new datasets the combined dataset is balanced, so:
1. The **Train Set** will have 80% of data.
2. The **Test Set** will have 20% of data and additionally more `spam` messages from 2021 and so on for future testing.

```python
# Random Shuffle
df_shuffled = df_comb.sample(frac=1, random_state=58).reset_index(drop=True)
df_shuffled
```

```python
df_spam = df_shuffled[df_shuffled['v1'] == "spam"].reset_index(drop=True)
df_spam
```

```python
df_ham = df_shuffled[df_shuffled['v1'] == "ham"].reset_index(drop=True)
df_ham
```

### Train Dataset

```python
import math

def len_80_percent_finder(df: pd.DataFrame) -> int:
    size = len(df)
    return math.ceil(size * (8/10))
```

```python
# Setting Up Training Dataset for balanced values
df_spam_80_per = df_spam.iloc[:len_80_percent_finder(df_spam)]
df_ham_80_per = df_ham.iloc[:len_80_percent_finder(df_ham)]

df_train = pd.concat([df_spam_80_per, df_ham_80_per], axis=0, ignore_index=True)
df_train = df_train.sample(frac=1, random_state=42).reset_index(drop=True)
df_train = pd.get_dummies(df_train, columns=["v1"], dtype=int)
df_train.columns = ["Text", "Ham", "Spam"]
df_train
```

```python
df_train['Ham'].value_counts()
```

```python
df_train.to_csv("Prepared-Data/train-data.csv", index=False)
```

### Test Dataset
- Adding Extra 2021 Tweets also (which are not added in train) to broadly test model performance.
- Source: https://github.com/opmusic/SpamHunter_dataset/blob/main/sms_spam_texts/2021_tweet_sms.zip
- Then adding 20% of combined dataset for testing

```python
import os
data_path = "Data/2021_tweet_sms_spam"
files_list_2021 = os.listdir(data_path)
files_list_2021
```

```python
spam_message_list_test: list[str] = []

for file in files_list_2021:
    
    with open(os.path.join(data_path, file)) as fp:
        data = fp.readlines()
        spam_message_list_test.extend([line.strip() for line in data])
        
        
test_spam_2021 = pd.DataFrame(spam_message_list_test, columns=["v2"])
test_spam_2021["v1"] = "spam"
test_spam_2021
```

```python
# Setting up test dataset as per planning
df_spam_20_per = df_spam.iloc[len_80_percent_finder(df_spam):]
df_ham_20_per = df_ham.iloc[len_80_percent_finder(df_ham):]

df_test = pd.concat([df_spam_20_per, df_ham_20_per, test_spam_2021], axis=0, ignore_index=True)
df_test = df_test.sample(frac=1, random_state=42).reset_index(drop=True)
df_test = pd.get_dummies(df_test, columns=["v1"], dtype=int)
df_test.columns = ["Text", "Ham", "Spam"]
df_test
```

```python
df_test["Spam"].value_counts()
```

```python
df_test.to_csv("Prepared-Data/test-data.csv", index=False)
```
